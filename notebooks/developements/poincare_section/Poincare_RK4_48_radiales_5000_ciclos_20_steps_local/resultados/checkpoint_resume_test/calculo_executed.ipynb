# Classical RK4 Poincare study

48 original radial particles, 5,000 forcing cycles, 20 fixed steps per cycle and four local processes. Original positions, IDs and colors are retained. The potential is the unchanged HDF5 mean plus first positive-frequency mode, B=1.5 T, characteristic length 0.06 m, source modes (0,1), cubic interpolation, rho=0.3. Normalized cycle duration is one. Arithmetic is float64. No reference or trajectory-accuracy certification is requested.

Classical RK4 is explicit and non-symplectic, with four field evaluations per step and no Newton solves or multiplier. Each process advances twelve independent particles. Immutable checkpoints store every step. The frozen public RK4 integrator runs fixed chunks, so restarted and uninterrupted chunked calculations use identical clocks. Floating-point time evaluation can differ slightly from a single unchunked call; validation checks that discrepancy on a short run. This focused method comparison follows the user-selected particle count, horizon and lack of reference rather than the project's default three-particle comparison protocol.
