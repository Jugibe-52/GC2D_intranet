# Generalized-energy conservation by step size

For a time-dependent Hamiltonian, the conserved extended quantity is
$K(t)=H(t,z(t))+k(t)$, where $\dot{k}=-\partial_t H$. This experiment compares
its relative error for the refinement sequence `0.004`, `0.001`, `0.00025`,
and `0.0000625`.


```python
import numpy as np

from workflows import (
    GeneralizedEnergyConfig,
    RandomPotentialConfig,
    centered_gc_trajectory,
    run_generalized_energy_comparison,
)
```

```python
potential_config = RandomPotentialConfig(
    amplitude=0.7,
    max_wave_number=25,
    nx=64,
    ny=64,
    seed=27,
    interpolation_order=5,
)
potential = potential_config.build()
trajectory = centered_gc_trajectory(potential, rho=0.3)
energy_config = GeneralizedEnergyConfig(
    steps=(4e-3, 1e-3, 2.5e-4, 6.25e-5),
    t_span=(0.0, 6 * np.pi),
    output_sample_count=361,
    coupling_frequency=np.pi / 8,
    progress=True,
)

print(potential_config)
print(energy_config)
```

## Integrations

The workflow builds the GC system, enables extended-momentum tracking, and
computes both absolute and relative generalized-energy errors for every step.


```python
result = run_generalized_energy_comparison(
    potential,
    trajectory,
    config=energy_config,
)
result.print_summary()
```
