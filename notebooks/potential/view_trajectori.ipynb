```python
import matplotlib.pyplot as plt
import numpy as np
from IPython.display import HTML
from pyhamsys import solve_ivp_sympext

from config_logging import configure_logging
from classes import Potential, PotentialSystem
from workflows.potentials import mock_potential

configure_logging()
```

```python
pot: Potential = mock_potential(A=0.7, M=25, nx=64, ny=64, seed=27, k=5)
print(pot)
```

```python
gc_system = PotentialSystem(pot, {'type': 'gc', 'rho': 0.3, 'eta': 0})
```

```python
t_eval = np.linspace(0, 4 * np.pi, 240)
z0_single = np.array([3.0, 4.0])
single_solution = solve_ivp_sympext(
    gc_system,
    (0, float(t_eval[-1])),
    z0_single,
    step=0.05,
    t_eval=t_eval,
    method='BM4',
    check_energy=False,  # El Hamiltoniano depende explicitamente de t.
)
```

```python
animated_electric_psi_trajectories = gc_system.animate_electric_psi_trajectories(
    single_solution,
    frames=80,
    step=1,
)
HTML(animated_electric_psi_trajectories.to_jshtml())
```
